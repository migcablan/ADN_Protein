## Ejercicio1: Replicacion del ADN

```python
from Bio.Seq import Seq

hebra_original = Seq("ATGCCGTTAGCT")

hebra_complementaria = hebra_original.complement()

hebra_rev_comp = hebra_original.reverse_complement()

print("=" * 50)
print("EJERCICIO 1: REPLICACIÓN DEL ADN")
print("=" * 50)
print(f"Hebra molde parental (5' -> 3'):     {hebra_original}")
print(f"Hebra complementaria  (3' <- 5'):     {hebra_complementaria}")
print(f"Hebra complementaria  (5' -> 3'):     {hebra_rev_comp}")

esperado_manual = "TACGGCAATCGA"
assert str(hebra_complementaria) == esperado_manual, "¡Error en la complementariedad!"
print("\n[OK] El resultado en Biopython coincide exactamente con el cálculo manual.")
```

## Ejercicio2: Transcripción del ADN a ARN

```python
from Bio import SeqIO
from Bio.Seq import Seq

nombre_archivo = "secuencia_ejercicio2.fasta"
with open(nombre_archivo, "w") as f:
    f.write(">gen_ejemplo_transcripcion\nATGCCTGAATGC\n")

registro = SeqIO.read(nombre_archivo, "fasta")
adn_codificante = registro.seq

arnm = adn_codificante.transcribe()

print("=" * 50)
print("EJERCICIO 2: TRANSCRIPCIÓN A ARNm")
print("=" * 50)
print(f"ID del registro FASTA:          {registro.id}")
print(f"Hebra codificante ADN (5'->3'): {adn_codificante}")
print(f"Hebra molde ADN       (3'<-5'): {adn_codificante.complement()}")
print(f"Transcrito de ARNm    (5'->3'): {arnm}")

arnm_hebra_opuesta = adn_codificante.reverse_complement().transcribe()
print(f"\n[Experimento] Si se transcribiera la otra hebra (5'->3'): {arnm_hebra_opuesta}")
print("Nota: Cambiar de hebra produce un ARNm completamente diferente y con distinta pauta de lectura.")
```

## Ejercicio3: Traducción del ARNm a proteína

```python
from Bio.Seq import Seq

arnm = Seq("AUGUAUGCUUAA")

traduccion_completa = arnm.translate(to_stop=False)
peptido = arnm.translate(to_stop=True)

print("=" * 50)
print("EJERCICIO 3: TRADUCCIÓN A PROTEÍNA")
print("=" * 50)
print(f"Secuencia ARNm:                {arnm}")
print(f"Traducción con símbolo STOP:   {traduccion_completa}")
print(f"Cadena de aminoácidos final:   {peptido} (Met-Tyr-Ala)")

arnm_mut_inicio = Seq("GUGUAUGCUUAA")
peptido_mut_inicio = arnm_mut_inicio.translate(to_stop=True)
print(f"\n[Mutación 1] Codón inicio AUG -> GUG:")
print(f"   ARNm: {arnm_mut_inicio} -> Traducido: {peptido_mut_inicio}")
print("   Efecto: En la célula impediría el ensamblaje correcto del ribosoma.")

arnm_sin_paro = Seq("AUGUAUGCUCAA") + Seq("GCCACCGGUUAA")
peptido_elongado = arnm_sin_paro.translate(to_stop=True)
print(f"\n[Mutación 2] Pérdida de codón STOP (UAA -> CAA):")
print(f"   ARNm elongado: {arnm_sin_paro}")
print(f"   Péptido resultante: {peptido_elongado} (elongación anómala)")
```

## Ejercicio6: Actividad integradora: Pipeline del Dogma Central

```python
from Bio import SeqIO
from Bio.Seq import Seq

def pipeline_dogma_central(archivo_fasta):
    """
    Lee un archivo FASTA de ADN e informa de cada etapa:
    1. Replicación (hebra molde complementaria)
    2. Transcripción (ARNm)
    3. Traducción (cadena peptídica)
    """
    print("=" * 65)
    print("       PIPELINE INTEGRAL: DEL ADN A LA PROTEÍNA")
    print("=" * 65)
    
    registro = SeqIO.read(archivo_fasta, "fasta")
    adn_codificante = registro.seq.upper()
    
    print(f"[*] Archivo cargado:         {archivo_fasta}")
    print(f"[*] Identificador:           {registro.id}")
    print(f"[*] Longitud de la cadena:   {len(adn_codificante)} pb\n")
    
    print("--- [PASO 1: REPLICACIÓN] ---")
    print("-> La helicasa abre la doble hélice y la ADN polimerasa replica.")
    dna_molde = adn_codificante.complement()
    print(f"Hebra codificante (5' -> 3'): {adn_codificante}")
    print(f"Hebra molde       (3' <- 5'): {dna_molde}\n")
    
    print("--- [PASO 2: TRANSCRIPCIÓN] ---")
    print("-> La ARN polimerasa transcribe el ADN a ARNm sustituyendo T por U.")
    arnm = adn_codificante.transcribe()
    print(f"Transcrito ARNm   (5' -> 3'): {arnm}\n")
    
    print("--- [PASO 3: TRADUCCIÓN] ---")
    print("-> El ribosoma decodifica los codones y ensambla los aminoácidos.")
    proteina_activa = arnm.translate(to_stop=True)
    proteina_completa = arnm.translate(to_stop=False)
    print(f"Traducción con símbolo STOP:  {proteina_completa}")
    print(f"Péptido funcional sintetizado: {proteina_activa}")
    print(f"Número de aminoácidos:        {len(proteina_activa)}")
    print("=" * 65)

fasta_pipeline = "pipeline_test.fasta"
with open(fasta_pipeline, "w") as f:
    f.write(">gen_ejemplo_pipeline\nATGCCGTTAGCTATGCCTGAATGCATGTAUGCTTAA\n".replace("U", "T"))

pipeline_dogma_central(fasta_pipeline)
```
